# **Import**

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
from data_loader import Dataset, standardize
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

# **Build From Scratch**

```python
class CustomLinearRegression():

    def __init__(self, learning_rate=0.001, num_iterations=10000):

        self.learning_rate = learning_rate
        self.num_iterations = num_iterations

    def fit(self, X, y):

        n, m = X.shape

        self.weights = np.random.randn(m)
        self.bias = np.random.randn()

        for _ in range(self.num_iterations):

            y_pred = self.predict(X)

            dw = (2 / n) * np.dot(X.T, (y_pred - y))
            db = (2 / n) * np.sum(y_pred - y)

            self.weights -= self.learning_rate * dw
            self.bias -= self.learning_rate * db

    def predict(self, X):

        return np.dot(X, self.weights) + self.bias
```

# **Load, Split and Standardize**

```python
dataset = Dataset("regression")
X_train, X_test, y_train, y_test = dataset.load_split_data()
X_train, X_test = standardize(X_train, X_test)
```

# **Train, Test and Compare**

```python
custom_model = CustomLinearRegression()
custom_model.fit(X_train, y_train)
y_pred_custom = custom_model.predict(X_test)

print(f"Custom Linear Regression MSE: {mean_squared_error(y_pred_custom, y_test):.3f}")

sklearn_model = LinearRegression()
sklearn_model.fit(X_train, y_train)
y_pred_sklearn = sklearn_model.predict(X_test)

print(f"Scikit-learn Linear Regression MSE: {mean_squared_error(y_pred_sklearn, y_test):.3f}")
```
